# Python notebook to format the table into LaTeX

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("params.csv")
```

```python
df
```

```python
def format_confidence_interval(vals, precision=3):
    """
    vals = [center, sd]
    precision = digits after decimal for coefficients
    """
    c, sd = vals

    # Determine a common exponent
    if c == 0:
        exp = 0
    else:
        exp = int(np.floor(np.log10(abs(c))))

    # Convert numbers to pseudo scientific notation with same exponent
    coeff_c = c / 10**exp
    coeff_sd = sd / 10**exp

    # Format numbers
    fmt = f"{{:.{precision}f}}"
    c_str = fmt.format(coeff_c)
    sd_str = fmt.format(coeff_sd)

    # Exponent formatting
    if exp == 0:
        return fr"${c_str} \pm {{{sd_str}}}$"
    else:
        return fr"$\left({c_str} \pm {{{sd_str}}}\right) \times 10^{{{exp}}}$"

def format_number(vals, precision=3):
    """
    vals = center
    precision = digits after decimal for coefficients
    """
    c = vals

    # Determine a common exponent
    if c == 0:
        exp = 0
    else:
        exp = int(np.floor(np.log10(abs(c))))

    # Convert numbers to pseudo scientific notation with same exponent
    coeff_c = c / 10**exp

    # Format numbers
    fmt = f"{{:.{precision}f}}"
    c_str = fmt.format(coeff_c)

    # Exponent formatting
    if exp == 0:
        return fr"${c_str}$"
    else:
        return fr"${c_str} \times 10^{{{exp}}}$"

```

```python
format_indices = [8, 10, 11, 12, 14]

tt_precisions = [
    1, 1, 1,
    2, 1, 1,
    2, 1,
    0,
    7,
    0, 1, 1,
    5,
    0,
    3
]

pott_precisions = [
    2, 3, 2,
    5, 4, 3,
    5, 3,
    2,
    8,
    2, 3, 2,
    6,
    0,
    7
]

truth_precisions = [
    5, 4, 6,
    4, 4, 4,
    4, 4,
    4,
    7,
    2, 2, 2,
    6,
    3,
    7
]

for i, row in df.iterrows():
    name = row.iloc[0]
    if i in format_indices:
        truth_str = format_number(row.iloc[1], truth_precisions[i])
    else:
        truth_str = str(np.round(row.iloc[1], truth_precisions[i]))
    tt_str = format_confidence_interval(np.array(row[4:6], dtype=np.float64), tt_precisions[i])
    pott_str = format_confidence_interval(np.array(row[2:4], dtype=np.float64), pott_precisions[i])

    main_str = " & ".join([name, truth_str, tt_str, pott_str])
    print(" ".join([main_str, "\\\\"]))
```
